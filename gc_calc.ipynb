```python
from Bio import SeqIO
import os


if not os.path.exists("results"):
    os.makedirs("results")

folder_path = "."

print("--- GC% Calculator ---")
print()

for filename in os.listdir(folder_path):
    if filename.endswith(".fna"):
        filepath = os.path.join(folder_path, filename)
        record = SeqIO.read(filepath, "fasta")
        seq = str(record.seq).upper()

        gc_count = seq.count('G') + seq.count('C')
        total = len(seq)
        gc_percent = (gc_count / total) * 100

        print(f"{filename} : {gc_percent:.2f}% GC")
```

```python
import os
import matplotlib.pyplot as plt


if not os.path.exists("results"):
    os.makedirs("results")

# Data
names = ['Bacillus', 'E.coli', 'Pseudomonas']
gc_values = [66.56, 50.79, 66.56]

# Bar Chart
plt.figure(figsize=(6,4))
plt.bar(names, gc_values, color=['blue', 'green', 'red'])
plt.ylabel('GC%')
plt.title('GC Content Comparison')
plt.ylim(0, 100)


plt.savefig('results/gc_bar.png')
plt.show()
```

```python
import matplotlib.pyplot as plt

# E.coli data
gc = 50.79
at = 49.21

plt.figure(figsize=(5,5))
plt.pie([gc, at], labels=['GC', 'AT'], autopct='%1.1f%%', colors=['orange', 'lightblue'])
plt.title('E.coli GC vs AT Content')
plt.savefig('results/ecoli_pie.png')
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,2))
ax.axis('off')

table_data = [
    ['Bacillus', '66.56%'], 
    ['E.coli', '50.79%'], 
    ['Pseudomonas', '66.56%']
]

table = ax.table(cellText=table_data, colLabels=['Bacteria', 'GC%'], loc='center', cellLoc='center')
table.auto_set_font_size(False)
table.set_fontsize(14)
table.scale(1, 2)

plt.title('GC% Results Summary', fontweight='bold', pad=20)
plt.savefig('results/gc_table.png')
plt.show()
```

## 1. Introduction
GC content is the percentage of Guanine and Cytosine in a DNA sequence. It varies widely between bacterial species and affects genome stability, gene expression, and adaptation to environment. Bacteria with high GC% often live in high temperature or extreme conditions.

In this project, I analyzed GC content variation across 3 bacterial genomes downloaded from NCBI: Bacillus, E.coli, and Pseudomonas. The goal was to calculate GC% and compare differences between species.

## 2. Methods
*Data Source:* Bacterial genome .fna files from NCBI
*Tools:* Python, Biopython, Matplotlib, Pandas
*Analysis Steps:*
1. Read genome sequences
2. Count G, C, A, T bases
3. Calculate GC% = [(G+C)/(A+T+G+C)] * 100
4. Visualize using bar chart, pie chart, and table
